```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("zomato_data.csv", encoding="latin1")

print(df.shape)
print(df.head())
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.dtypes)
```

```python
df
```

```python
print(df.describe())
```

```python
city_count = df['City'].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=city_count.index,
    y=city_count.values
)

plt.title("top 10 cities by number of restaurants")
plt.xlabel("city")
plt.ylabel("number of restaurants")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Aggregate rating',
    bins=20,
    kde=True
)

plt.title("distribution of restaurant ratings")
plt.xlabel("aggregate rating")
plt.ylabel("number of restaurants")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Price range'
)

plt.title("number of restaurants by price range")
plt.xlabel("price range")
plt.ylabel("number of restaurants")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Votes',
    y='Aggregate rating'
)

plt.title("votes vs restaurant rating")
plt.xlabel("number of votes")
plt.ylabel("aggregate rating")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='Price range',
    y='Aggregate rating'
)

plt.title("restaurant rating by price range")
plt.xlabel("price range")
plt.ylabel("aggregate rating")
plt.show()
```

```python
avg_rating = df.groupby('Price range')['Aggregate rating'].mean()

plt.figure(figsize=(8, 5))

plt.plot(
    avg_rating.index,
    avg_rating.values,
    marker='o'
)

plt.title("average rating by price range")
plt.xlabel("price range")
plt.ylabel("average rating")
plt.grid()
plt.show()
```

```python
delivery_count = df['Has Online delivery'].value_counts()

plt.figure(figsize=(6, 5))

sns.barplot(
    x=delivery_count.index,
    y=delivery_count.values
)

plt.title("restaurants offering online delivery")
plt.xlabel("online delivery")
plt.ylabel("number of restaurants")
plt.show()
```

```python
numeric_columns = [
    'Average Cost for two',
    'Price range',
    'Aggregate rating',
    'Votes'
]

plt.figure(figsize=(8, 6))

sns.heatmap(
    df[numeric_columns].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("correlation between restaurant variables")
plt.show()
```
