```python
#%pip install matplotlib seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
df = pd.read_csv("Coffe_sales.csv")

print(df.shape)
print(df.isnull().sum())
print("dupli =",df.duplicated().sum())
```

```python
print(df.dtypes)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python

coffee_count = df['coffee_name'].value_counts()

plt.figure(figsize=(10, 5))
sns.barplot(x=coffee_count.index, y=coffee_count.values)

plt.xticks(rotation=45)
plt.xlabel("coffee type")
plt.ylabel("number of sales")
plt.title("number of sales by coffee type")
plt.show()
```

```python
daily_sales = df.groupby('Date')['money'].sum()

plt.figure(figsize=(12, 5))
plt.plot(daily_sales.index, daily_sales.values)

plt.xlabel("date")
plt.ylabel("total sales")
plt.title("daily coffee sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['money'], bins=20, kde=True)

plt.xlabel("amount")
plt.ylabel("frequency")
plt.title("distribution of transaction amount")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='coffee_name', y='money')

plt.xticks(rotation=45)
plt.xlabel("coffee type")
plt.ylabel("amount")
plt.title("transaction amount by coffee type")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(8, 5))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')

plt.title("correlation between numerical variables")
plt.show()
```
