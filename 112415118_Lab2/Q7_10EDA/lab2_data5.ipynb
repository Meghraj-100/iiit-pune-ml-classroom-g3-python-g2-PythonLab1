```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("pokemon.csv")
print(df.shape)
df
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.dtypes)
```

```python
df.drop('evolves_from',axis=1 , inplace=True,)
df
```

```python
print(df.isnull().sum())
```

```python
df['total_stats'] = (
    df['hp'] +
    df['attack'] +
    df['defense'] +
    df['special_attack'] +
    df['special_defense'] +
    df['speed']
)
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='generation')

plt.title("number of pokemon by generation")
plt.xlabel("generation")
plt.ylabel("number of pokemon")
plt.show()
```

```python
legendary_count = df['is_legendary'].value_counts()

plt.figure(figsize=(6, 5))

sns.barplot(
    x=legendary_count.index,
    y=legendary_count.values
)

plt.title("legendary pokemon")
plt.xlabel("is legendary")
plt.ylabel("number of pokemon")
plt.show()
```

```python
mythical_count = df['is_mythical'].value_counts()

plt.figure(figsize=(6, 5))

sns.barplot(
    x=mythical_count.index,
    y=mythical_count.values
)

plt.title("mythical pokemon")
plt.xlabel("is mythical")
plt.ylabel("number of pokemon")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.countplot(
    data=df,
    x='category'
)

plt.title("number of pokemon by category")
plt.xlabel("category")
plt.ylabel("number of pokemon")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='habitat'
)

plt.title("number of pokemon by habitat")
plt.xlabel("habitat")
plt.ylabel("number of pokemon")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='hp',
    bins=20,
    kde=True
)

plt.title("distribution of pokemon hp")
plt.xlabel("hp")
plt.ylabel("frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='attack',
    y='defense',
    hue='is_legendary'
)

plt.title("attack vs defense")
plt.xlabel("attack")
plt.ylabel("defense")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='height',
    y='weight',
    hue='is_legendary'
)

plt.title("height vs weight")
plt.xlabel("height")
plt.ylabel("weight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='speed',
    bins=20,
    kde=True
)

plt.title("distribution of pokemon speed")
plt.xlabel("speed")
plt.ylabel("frequency")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x='is_legendary',
    y='total_stats'
)

plt.title("total stats of legendary and non-legendary pokemon")
plt.xlabel("is legendary")
plt.ylabel("total stats")
plt.show()
```

```python
stats = [
    'hp',
    'attack',
    'defense',
    'special_attack',
    'special_defense',
    'speed',
    'height',
    'weight'
]

plt.figure(figsize=(10, 8))

sns.heatmap(
    df[stats].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("correlation between pokemon characteristics")
plt.show()
```

```python
top10 = df.sort_values(
    'total_stats',
    ascending=False
).head(10)

print(
    top10[['name', 'total_stats', 'is_legendary', 'is_mythical']]
)
```
