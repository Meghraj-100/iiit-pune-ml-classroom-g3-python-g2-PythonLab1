```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('bike_hour.csv')
df
```

```python
print(df.columns)
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
print(df.describe())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(figsize=(12, 8), bins=20)

plt.suptitle("distribution of numerical variables")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("correlation between numerical variables")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=df[numeric_cols])

plt.title("boxplot of numerical variables")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x='season',
    y='cnt'
)

plt.title("average bike rentals by season")
plt.xlabel("season")
plt.ylabel("average rentals")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='temp',
    y='cnt'
)

plt.title("temperature vs bike rentals")
plt.xlabel("temperature")
plt.ylabel("number of rentals")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='hum',
    y='cnt'
)

plt.title("humidity vs bike rentals")
plt.xlabel("humidity")
plt.ylabel("number of rentals")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='cnt', bins=30, kde=True)

plt.title("distribution of bike rentals")
plt.xlabel("number of rentals")
plt.ylabel("frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='season', y='cnt')

plt.title("bike rentals by season")
plt.xlabel("season")
plt.ylabel("number of rentals")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

corr = df[['temp', 'atemp', 'hum', 'windspeed',
           'casual', 'registered', 'cnt']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title("correlation heatmap of bike sharing data")
plt.show()
```
