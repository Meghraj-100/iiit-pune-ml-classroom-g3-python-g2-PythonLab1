## Strings in Python

Open the file poem.txt and print lenght of each line.

Find sum of each line, mean, median and standard deviation of all lines in the file.

Count the number of words in the file

Print the first 1000 characters from poem.txt.

Print the 3rd to 5th words from the file.

```python
import statistics

with open('poem.txt', 'r') as file:
    lines = file.readlines()

length = []
count = 0

for line in lines:
    stripped = line.strip()
    print(len(stripped))
    length.append(len(stripped))
    count += len(stripped.split())

text = ''.join(lines)
print(text[:1000])

words = text.split()
print(words[2:5])

print("Total Words:", count)
print("Sum:", sum(length))
print("Mean:", statistics.mean(length))
print("Median:", statistics.median(length))
print("Standard Deviation:", statistics.stdev(length))
```

# Working with CSV

Let's import a file with fuel economy data

```python
import csv
with open('mpg.csv') as csvfile:
     mpg = list(csv.DictReader(csvfile))
    
```

Find out how many items are there in the dictionary

```python
for item in mpg[1].items():
    print(item)

print("number of items:", len(mpg[1]))

print(mpg[0].keys())

total = 0

for item in mpg:
    total += float(item['cty'])

average_cty = total / len(mpg)

print("average city fuel economy:", average_cty)

total = 0

for item in mpg:
    total += float(item['hwy'])

average_hwy = total / len(mpg)

print("average highway fuel economy:", average_hwy)

```

Find the column names with keys

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

Find the average hwy fuel economy across all cars

Use set to return the unique values for the number of cylinders the cars.

```python
cylinders = set(d['cyl'] for d in mpg)
cylinders
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
mpgByCyl = []

for c in cylinders:
    # iterate over all the cylinder levels
    summpg = 0
    typecount = 0

    for d in mpg:
        # iterate over all dictionaries
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1

    mpgByCyl.append((c, summpg / typecount))

mpgByCyl.sort(key=lambda x: x[0])
mpgByCyl
```

Return the unique values for the class types in our dataset using set.

```python
vehicleclass = set(d['class'] for d in mpg)
```

Find average miles per gallon in the highway for each class of vehicles

```python
MileageByClass = []

for t in vehicleclass:
    # iterate over all the vehicle classes
    summpg = 0
    classCount = 0

    for d in mpg:
        if d['class'] == t:
            summpg += float(d['hwy'])
            classCount += 1

    MileageByClass.append((t, summpg / classCount))

MileageByClass.sort(key=lambda x: x[1])

MileageByClass
```
